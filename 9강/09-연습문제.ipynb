분포 형태 분석: 평균이 중앙값보다 완전 크므로, 완전 우측으로 치우친 분포로 볼 수 있다.

#### 2-3. 박스플롯을 그려 이상값의 존재 여부를 확인하세요

```python
result = []
for f in fields:
    mean_val = df[f].mean()
    median_val = df[f].median()

    if mean_val > median_val:
        skew_type = "Right-skewed"
    elif mean_val < median_val:
        skew_type = "Left-skewed"
    else:
        skew_type = "Symmetric"
    result.append({
        'field': f, 'mean': mean_val, 'median': median_val, 'skew':
        skew_type
    })
DataFrame(result)
```

```python
# 1) 그래프 초기화 (캔바스(fig)와 도화지(ax) 준비하기)
width_px = 2000
height_px = 600
rows = 1
cols = 5
figsize = (width_px / my_dpi, height_px / my_dpi)
fig, ax = plt.subplots(rows, cols, figsize=figsize, dpi=my_dpi)


# 2) 그래프 그리기
for i, v in enumerate(fields):
 sb.boxplot(data=df, y=v, ax=ax[i])
 ax[i].set_title(v)
 ax[i].grid(alpha=0.3)



plt.tight_layout()
#plt.savefig("myplot.png", dpi=my_dpi)
plt.show() # 그래프 화면 출력
plt.close()
```

이상값이 관측되는 변수가 존재한다.
이상치에 대한 처리는 분석가의 주관이 반영될 수 있기 때문에 이상치를 처리할 것인지, 그대로 둘
것인지는 분석가의 판단에 따른다.
만약, 이상치를 그대로 둔다면 이상치에 민감한 분석 기법들이 많기 때문에 스케일링, 로그변환 등
으로 이상치에 대한 민감성을 축소시켜야 한다. 

### 📝 Mission 3: 분포 정규성 확인

#### 3-1. 주택 가격(MEDV)의 히스토그램을 그리고 분포의 형태를 분석하세요.

```python
# 1) 그래프 초기화 (캔바스(fig)와 도화지(ax) 준비하기)
width_px = 1000
height_px = 800
rows = 1
cols = 1
figsize = (width_px / my_dpi, height_px / my_dpi)
fig, ax = plt.subplots(rows, cols, figsize=figsize, dpi=my_dpi)

#2) 그래프 그리기 -> seaborn 사용
# 2) Histogram 그리기
sb.histplot(data=df, x="MEDV", bins=7, kde=True,
edgecolor="#000000", linewidth=0.5)

#3) 그래프 꾸미기
ax.grid(True)
# 4) 출력

plt.tight_layout()
#plt.savefig("myplot.png", dpi=my_dpi)
plt.show() # 그래프 화면 출력
plt.close()
```

분포는 오른쪽 꼬리가 긴 형태를 보인다. (high-value outlier 또는 비대칭적 분포)
평균(22.53)이 중앙값(21.2)보다 크다, 이는 오른쪽으로 치우친(right-skewed) 분포의 전형적
특징이다.
대다수의 값은 약 15~25 사이에 집중되어 있으며, 히스토그램의 가장 높은 구간도 이 근처에
위치한다.
30 이상 고가(MEDV≥30) 구간의 빈도는 상대적으로 낮다, 즉 고가 지역은 소수다.
KDE 곡선의 최고점(모드)이 20 근처에 있어 최빈값은 20 전후로 관측된다.
50 근처에서 관측되는 값은 매우 드물다, 즉 극단적 고가 값은 소수의 사례다.
10 이하의 값도 비교적 적다, 즉 매우 저가 지역 또한 상대적으로 적다.
분포 전체는 한쪽 꼬리가 길고, 중심부의 밀도가 뚜렷하며, 극단값(outliers)이 존재하는 전형
적인 부동산 가격 패턴을 보여준다. 

#### 3-2. 심하게 치우친 변수들(CRIM, NOX, LSTAT)에 대해 로그 변환을 적용하고,    변환 전후의 분포를 비교분석하세요

```python
skewed_vars = ['CRIM', 'NOX', 'LSTAT']

# 1) 그래프 초기화
width_px = 2000
height_px = 2000
rows = len(skewed_vars)
cols = 2
figsize = (width_px / my_dpi, height_px / my_dpi)
fig, ax = plt.subplots(rows, cols, figsize=figsize, dpi=my_dpi)


for i, v in enumerate(skewed_vars):
    # 원본에 대한 분포 확인
    sb.histplot(data=df, x=v, bins=10, kde=True, ax=ax[i][0])
    ax[i][0].set_title(f'{v} - 원본')
    ax[i][0].grid(alpha=0.3)
    # log변환 후의 분포 확인
    log_data = np.log10(df[v] + 1)
    sb.histplot(data=log_data, bins=10, kde=True, ax=ax[i][1])
    ax[i][1].set_title(f'{v} - 로그변환')
    ax[i][1].grid(alpha=0.3)



plt.tight_layout()
plt.show()
plt.close()
```
